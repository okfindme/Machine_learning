GD Complexity:
GD's complexity is O(n × iterations), where n is the number of features and iterations is the number of steps to reach the minimum. In each step, GD computes a slope for every single parameter — called a partial derivative — which tells each parameter whether to increase or decrease to go downhill. This means one step isn't one calculation, it's n calculations. Learning rate doesn't appear in the formula directly, but it indirectly affects complexity — a smaller learning rate means more iterations to converge, making the algorithm slower overall. Compared to Normal Equation's O(n³), GD scales far more efficiently on high dimensional data.

### Local Minima & GD Variants:

In Gradient Descent, a local minima is a point where the algorithm thinks it has found the lowest point, but there is actually a deeper valley (global minima) somewhere else. Batch GD uses all rows of data every step, so its slope calculation is very accurate and always points cleanly downhill — meaning once it gets trapped in a local minima, it never escapes. SGD (Stochastic Gradient Descent) fixes this by using only 1 random row per step, making the slope noisy and inaccurate. This noise occasionally points the algorithm accidentally uphill, which is exactly what kicks it out of a local minima valley. Mini-Batch GD is the middle ground — you choose a batch size (like 32 or 64 rows) per step, balancing noise and accuracy. Important: "one sample" in SGD means one row (one complete observation), not one column/feature. Columns are features, rows are individual data points.

### Closed form 
direct formula, one shot answer. Example: Normal Equation.
### No closed form 
no direct formula exists, so you iterate your way to the answer. Example: GD in Neural Networks.
