```python
import numpy as np 
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
```

```python
x , y = load_diabetes(return_X_y= True)
x_train, x_test,y_train,y_test = train_test_split(x, y , test_size=0.3 , random_state=42)
```

```python
class MiniBatch():
    def __init__(self , lr = 0.01 , epoches = 50):
        self.lr = lr  
        self.epoches = epoches 
        self.m = None
        self.b = 0 

    def fit(self , x_train , y_train , batch_size) :
        self.m = np.ones(x_train.shape[1])
        x = x_train
        y = y_train 
        
        no_batch = y_train.shape[0] / batch_size
        
        for num in range(self.epoches):

            for iterate in range(int(no_batch)): # so that if no_of batch comes in the float it gets converted into the integer

                random_rows = random.sample(range(x_train.shape[0]), batch_size)

                # as now shape of x is ( batch_size , 10 ) and m shape is (10,) which means 10 individual values 
                # hence we cannot perform the normal mutliplication else it will give result of diff rows , in dot multiplication result will be in singal matric ( batch_size ,)
                y_hat = np.dot(x[random_rows], self.m) + self.b

                # in this y[random_rows] has 32 values means it has shape of ( 32 ,)
                # x will be in ( 32 , 10 )
                # so np.dot transform the y in ( 1 , 32 ) , dot product will be in ( 1,10)
                # as we are finding gradient using the mean_square hence we will find its mean by divind it with the batch size 
                loss_slope_m = -2 * np.dot((y[random_rows] - y_hat), x[random_rows]) / batch_size
                loss_slope_b = -2 * np.mean(y[random_rows] - y_hat)

                self.m = self.m - (self.lr * loss_slope_m)
                self.b = self.b - (self.lr * loss_slope_b)

    def predict(self , x_test):
        return np.dot(x_test , self.m) +self.b

```
