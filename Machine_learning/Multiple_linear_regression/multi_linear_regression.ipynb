```python
from sklearn.datasets import make_regression as dummy_data
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

```python
"""
creating the dummy_data for multiple_linear_regression 
n_samples =  tells the number of rows that should be there 
n_feature =  tells the number of column that should be there 
n_informative = tells how many columns ( which we created ) will effect the target 
n_target tells =  how many output column should be there 
noise = tells how much error should be there , as when the algo will find the line/plane how much error it will have from the point
"""

# result unpacking
x,y = dummy_data(n_samples=1000,n_features=4, n_informative=3 , n_targets=1 , noise=70 )
```

```python
df = pd.DataFrame(x, columns=['column1', 'column2', 'column3', 'column4'])
df['target'] = y
```

```python
print(df)
```

# 3d ploting the graph 

as it is more easy to understand the data

```python
px.scatter_3d(df , x ='column1',y= 'column4', z= 'target' )
```

## performing the linear rigression

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
x_train , x_test , y_train , y_test = train_test_split(df[['column1', 'column2', 'column3', 'column4']],df[["target"]],test_size=0.3,random_state=42)
```

```python
lr = LinearRegression()
```

```python
lr.fit(x_train , y_train)
```

```python
lr.coef_ # column 1 has coef is near 0 means this column is no use
```

```python
y_pred = lr.predict(x_test)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_result = r2_score(y_true=y_test,y_pred=y_pred)
print(r2_result)
```
