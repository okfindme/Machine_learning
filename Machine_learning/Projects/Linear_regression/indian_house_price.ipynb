### now targeting the elephant of the room - aminities 

```python
df['Amenities'].value_counts()
```

```python
from sklearn.preprocessing import MultiLabelBinarizer

# Step 1 - convert string to list
df_delhi["Amenities"] = df_delhi["Amenities"].apply(lambda x: x.split(", "))

# Step 2 - apply MultiLabelBinarizer
mlb = MultiLabelBinarizer()
amenities_encoded = pd.DataFrame(
    mlb.fit_transform(df_delhi["Amenities"]),
    columns=mlb.classes_,
    index=df_delhi.index  # important to align indexes before concat!
)

# Step 3 - concat and drop original
df_delhi = pd.concat([df_delhi, amenities_encoded], axis=1)
df_delhi.drop("Amenities", axis=1, inplace=True)
```

## machine learnig algo 

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
x = df_delhi.drop(columns=['Price_in_Lakhs'])
y = df_delhi['Price_in_Lakhs']
```

```python
x_train , x_test , y_train , y_test = train_test_split(x, y , random_state=42 , test_size=0.3)
```

```python
lr = LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
r2_score(y_pred=y_pred,y_true=y_test)
```
