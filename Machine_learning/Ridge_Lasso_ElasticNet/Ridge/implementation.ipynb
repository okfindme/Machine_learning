```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
import numpy as np 
```

```python
x , y = load_diabetes(return_X_y= True)
x_train , x_test , y_train , y_test = train_test_split(x, y, test_size=0.2 , random_state=4)
```

```python
print("x_train = ",x_train )
print("x_test = ",x_test )
```

```python
print("X_train shape:", x_train.shape)
print("X_test shape:", x_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
from numpy import dot


class my_ridge ():
    def __init__(self, alpha = 0.1 ):
        self.alpha = alpha 
        self.coef_ = None
        self.intercept_ = None 

    def fit(self , x  , y):
        x = np.insert(x,0,1 , axis=1)
        I = np.identity(x.shape[1])
        I[0][0] = 0

        w =np.dot( np.linalg.inv(np.dot(x.T,x)+self.alpha*I), np.dot(x.T,y))
        self.intercept_ = w[0]
        self.coef_ = w[1:]

    def predict(self , x_test):
        return np.dot(x_test ,self.coef_ ) + self.intercept_
        
```

```python
rd = my_ridge(alpha=0.1)
rd.fit(x_train , y_train)

rd_pred = rd.predict(x_test)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test , rd_pred)
```

```python
from sklearn.linear_model import Ridge
# A larger alpha value means stronger regularization
ridge_model = Ridge(alpha=0.1)
ridge_model.fit(x_train, y_train)
predictions = ridge_model.predict(x_test)
score = r2_score( y_test, predictions) # R-squared score
print(f"R-2 score: {score}")
```
