## Exception_handling 
Exception handling is a mechanism used in programming to handle runtime errors (errors that occur while the program is running) so that the program does not crash suddenly and can continue or terminate gracefully.

```python
# try except block
try:
    a = b
except :
    print("variable is not assigned")
```

```python
try :
    a = 10/0
    a = b 
except Exception as ex : # Exception is the main class from which other classes are derived
    print(ex)
except ValueError as vr : # vlaueerror will only be trigerred if variable is not defined
    print(vr)
else : # else will only be excuted it no error occured
    print("no error occured")
```
